```python
import xml.dom.minidom
```

```python
#Creo un diccionario para que pueda servir de plantilla y simplificar el código. Es decir, para que no haya que meter el código a mano por cada nuevo arctículo. También se complica si el artículo tiene varios contributors, lo que es relativamente probable

articulo = {
    "tipo_articulo": "research-article", #tipos: research-article, review, editorial, letter, commentary, erratum, other
    "journal_id":"ARIS",
    "journal_title":"Arte, individuo y sociedad",
    "abbrev_journal_title":"Art., indv. y soc.",
    "issn":"meter issn",
    "issn_l":"meter issn-l",
    "isbn":"meter ISBN",
    "publisher_name":"Universidad Complutense de Madrid",
    "publisher_country":"España",
    "article_id":"meter url DOI",
    "subject_discipline":"Biología",
    "subject_section":"artículo",
    "article_title":"meter el título del artículo",
    "trans_title":"meter la traducción del título del artículo",
    "day":"meter día de publicación",
    "month":"meter mes de publicación",
    "year":"meter año de publicación",
   #Fecha recibido
    "day_received":"02",
    "month_received":"08",
    "year_received":"2024",
    #Fecha de revisión
     "day_rev_recd":"06",
    "month_rev_recd":"03",
    "year_rev_recd":"2024",
    #Fecha de aceptación
    "day_accepted":"06",
    "month_accepted":"03",
    "year_accepted":"2026",
    #Resumen
    "resumen_español": "introduce resumen en español",
    #Abstract
    "abstract_ingles": "introduce abstract en inglés",
    #palabras clave
    "palabras_clave": ["palabras clave 1", "palabra clave 2", "palabra clave 3", "palabra clave 4", "palabra clave 5"], #añade, entre comillas y con una coma, y dentro del corchete, más palabras clave
    #keywords
    "keywords": ["keyword 1", "keyword 2", "keyword 3", "keyword 4", "keyword 5"], #añade, entre comillas y con una coma, y dentro del corchete, más palabras clave
    #sumario
    "sumario":"introduce el sumario",
    #citar
    "citar":"introduce el texto de cómo citar",


    #Vamos a meter varios contribuidores, porque es probable que algún artículo tenga varios autores o contruibuidores, metiendo una lista de diccionarios
    "contributors" : [
            {"contrib_id":"ORCID_1",
            "surname":"apellido_1",
            "given_names":"nombre_1",
            "institution": "Universidad_1",
            "country": "País_1",
            "aff_id": "aff-a",  # ID para la afiliación
            "corresp_id": "cor1",  # ID para la correspondencia
            "email":"Apellido_4@Universidad_1",
            },
            {"contrib_id":"ORCID_2",
            "surname":"Apellido_2",
            "given_names":"Nombre_2",
            "institution": "Universidad_2",
            "country": "País_2",
            "aff_id": "aff-a",
            "corresp_id": "cor2",
            "email":"Apellido_4@Universidad_2",
            },
            {"contrib_id":"ORCID_3",
            "surname":"Apellido_3",
            "given_names":"Nombre_3",
            "institution": "Universidad_3",
            "country": "País_3",
            "aff_id": "aff-c",
            "corresp_id": "cor3",
            "email":"Apellido_4@Universidad_3",
            },
            {"contrib_id":"ORCID_4",
            "surname":"Apellido_4",
            "given_names":"Nombre_4",
            "institution": "Universidad_4",
            "country": "País_4",
            "aff_id": "aff-d",
            "corresp_id": "cor4",
            "email":"Apellido_4@Universidad_4",

            },
            {"contrib_id":"ORCID_5",
            "surname":"Apellido_5",
            "given_names":"Nombre_5",
            "institution": "Universidad_5",
            "country": "País_5",
            "aff_id": "aff-e",
            "corresp_id": "cor5",
            "email":"Apellido_5@Universidad_5"
            }],
            #si se quieren meter más contribuidores solo hay que copiar uno de los elementos del diccionario y rellenar los datos

           # Permissions
    "permissions": {
        "copyright_statement": "Copyright © 2024, Universidad Complutense de Madrid",
        "copyright_year": "2024",
        "copyright_holder": "Universidad Complutense de Madrid",
        "license_href": "https://creativecommons.org/licenses/by/4.0/",
        "license_ref": "https://creativecommons.org/licenses/by/4.0/",
        "license_p_text": "Esta obra está bajo una licencia ",
        "license_p_link_text": "Creative Commons Attribution 4.0 International"
    }

    }


```

```python
# Para que esto se pueda cumplir, hace falta nombrar los namespaces: <article article-type="research-article" dtd-version="1.3" xml:lang="es" xmlns:mml="http://www.w3.org/1998/Math/MathML" xmlns:xlink="http://www.w3.org/1999/xlink" xmlns:xsi="http://www.w3.org/2001/XMLSchema-instance" xmlns:ali="http://www.niso.org/schemas/ali/1.0/" xmlns:oasis="http://www.niso.org/standards/z39-96/ns/oasis-exchange/table">

ET.register_namespace("xlink", "http://www.w3.org/1999/xlink")
ET.register_namespace("ali", "http://www.niso.org/schemas/ali/1.0/")
ET.register_namespace("mml", "http://www.w3.org/1998/Math/MathML")
ET.register_namespace("xsi", "http://www.w3.org/2001/XMLSchema-instance")
ET.register_namespace("oasis", "http://www.niso.org/standards/z39-96/ns/oasis-exchange/table")
```

```python
#nivel 1 o raiz de JATS

raiz = ET.Element("article")
raiz.set("article-type", articulo["tipo_articulo"])
raiz.set("dtd-version", "1.3")
raiz.set("{http://www.w3.org/XML/1998/namespace}lang", "es")

#nivel 2: Front

front  = ET.SubElement(raiz,"front")

#nivel 3: elementos del Front

journal_meta = ET.SubElement(front, "journal-meta")
article_meta = ET.SubElement(front, "article-meta")

#nivel 4.1:elementos de journal-meta

journal_id = ET.SubElement(journal_meta, "journal-id", attrib={"journal-id-type": "publisher-id"})
journal_title_group = ET.SubElement(journal_meta, "journal-title-group")
#aquí debería haber lo de aff y contributor-group (no parece necesario en el journal-meta)
issn = ET.SubElement(journal_meta, "issn", attrib={"publication-format":"electronic"})
issn_l = ET.SubElement(journal_meta, "issn-l")
isbn = ET.SubElement(journal_meta, "isbn")
publisher = ET.SubElement(journal_meta, "publisher")

#nivel 4.1.1:hijos de la etiqueta journal-title-group

journal_title = ET.SubElement(journal_title_group, "journal-title", attrib={"specific-use": "original","xml:lang": "es"})
abbrev_journal_title = ET.SubElement(journal_title_group, "abbrev-journal-title", attrib={"abbrev-type":"publisher"})

#nivel 4.1.2:hijos de la etiqueta publisher

publisher_name = ET.SubElement(publisher, "publisher-name")
publisher_loc = ET.SubElement(publisher, "publisher-loc")

#nivel 4.1.2:hijos de la etiqueta publisher-loc

country = ET.SubElement(publisher_loc, "country")


#completar texto journal-meta (esto es lo que puede rellenar una persona sin conocimiento de python, pero lo vamos a hacer a través de un diccionario)

journal_id.text= articulo["journal_id"]
journal_title.text= articulo["journal_title"]
abbrev_journal_title.text= articulo["abbrev_journal_title"]
issn.text= articulo["issn"]
issn_l.text=articulo ["issn_l"]
isbn.text= articulo ["isbn"]
publisher_name.text= articulo["publisher_name"]
country.text= articulo["publisher_country"]

#nivel 4.2: elementos de article-metadata

article_id = ET.SubElement(article_meta, "article-id", attrib={"pub-id-type":"doi"})
article_categories = ET.SubElement(article_meta, "article-categories")
title_group =ET.SubElement(article_meta, "title-group")
contrib_group = ET.SubElement(article_meta, "contrib-group")
#A los hijos de contrib_group lo voy a meter más abajo en un bucle for

#nivel 4.2.1: hijos etiqueta article-categories

subj_group_discipline = ET.SubElement(article_categories,"subj-group", attrib={"subj-group-type":"discipline"})
subj_group_section = ET.SubElement(article_categories,"subj-group", attrib={"subj-group-type":"section"})

#nivel 4.2.1.1: hijos etiqueta subj-group

subject_discipline = ET.SubElement(subj_group_discipline, "subject")
subject_section = ET.SubElement(subj_group_section, "subject")

#nivel 4.2.2: hijos etiqueta title_group

article_title = ET.SubElement(title_group, "article-title", attrib={"xml:lang": "es"})
trans_title_group = ET.SubElement(title_group, "trans-title-group", attrib={"xml:lang": "en"})

#nivel 4.2.2.1: hijos etiqueta title_group

trans_title = ET.SubElement(trans_title_group, "trans-title")

#completar texto article-meta (esto es lo que puede rellenar una persona sin conocimiento de python, pero lo vamos a hacer a través de un diccionario)

article_id.text=articulo["article_id"]
subject_discipline.text=articulo["subject_discipline"]
subject_section.text=articulo['subject_section']
article_title.text=articulo["article_title"]
trans_title.text=articulo["trans_title"]

#Todo estolo vamos a hacer con bucles: etiqueta 4.2 de contrib y nivel 4.2.3: hijos contrib-group y #nivel 4.2.3.1: hijos de la etiqueta contrib y #nivel 4.2.3.1: hijos de la etiqueta contrib y #nivel 4.2.3.1.1: hijos de la etiqueta name

# pensar si ponemos el rol: role = ET.SubElement(contrib, "role")

for persona in articulo["contributors"]:
    contrib = ET.SubElement(contrib_group, "contrib", attrib={"contrib-type":"author", "corresp": "yes"})

    contrib_id = ET.SubElement(contrib, "contrib-id", attrib={"contrib-id-type":"orcid"})
    contrib_id.text = persona["contrib_id"]

    name = ET.SubElement(contrib, "name")
    surname = ET.SubElement(name, "surname")
    surname.text = persona["surname"]

    given_names = ET.SubElement(name, "given-names")
    given_names.text = persona["given_names"]

    xref_aff = ET.SubElement(contrib, "xref", attrib={"ref-type": "aff", "rid": persona["aff_id"]})
    xref_corresp = ET.SubElement(contrib, "xref", attrib={"ref-type": "corresp", "rid": persona["corresp_id"]})


# Para crear una etiqueta de afiliación

afiliaciones_unicas = {}

for persona in articulo["contributors"]:
    aff_id = persona["aff_id"]
    if aff_id not in afiliaciones_unicas:
        afiliaciones_unicas[aff_id] = {
            "institution": persona.get("institution", ""),
            "country": persona.get("country", "")
        }

# Para crear una etiqueta <aff> por cada aff_id único y que no se dupliquen
for aff_id, datos in afiliaciones_unicas.items():
    aff = ET.SubElement(contrib_group, "aff", attrib={"id": aff_id})
    institution = ET.SubElement(aff, "institution", attrib={"content-type":"original"})
    institution.text = datos["institution"]
    country = ET.SubElement(aff, "country")
    country.text = datos["country"]


# Para crear etiqueta author_notes

author_notes = ET.SubElement(article_meta, "author-notes")

for persona_email in articulo["contributors"]:
    correspondencia = ET.SubElement(author_notes, "corresp", attrib={"id": persona_email["corresp_id"]})
    correspondencia.text = f"{persona_email['given_names']} {persona_email['surname']}"
    email = ET.SubElement(correspondencia, "email")
    email.text = persona_email["email"]


#seguimos con los hijos del nivel 4.2: elementos de article-metadata

pub_date = ET.SubElement(article_meta, "pub-date", attrib={"date-type":"pub", "publication-format":"electronic", "iso-8601-date":"2025-03-06"})

#seguimos con los hijos del nivel 4.2.1: hijos de pub_date

day = ET.SubElement(pub_date, "day")
month = ET.SubElement(pub_date, "month")
year = ET.SubElement(pub_date, "year")
day.text = articulo["day"]
month.text = articulo["month"]
year.text = articulo["year"]


#seguimos con los hijos del nivel 4.2: elementos de article-metadata

history = ET.SubElement(article_meta, "history")

#seguimos con los hijos del nivel 4.2.1: elementos history

date_received = ET.SubElement(history, "date", attrib={"date-type":"received", "iso-8601-date":"2024-12-16"})
date_rev_recd = ET.SubElement(history, "date", attrib={"date-type":"rev-recd", "iso-8601-date":"2024-12-16"})
date_accepted = ET.SubElement(history, "date", attrib={"date-type":"accepted", "iso-8601-date":"2024-12-16"})

#seguimos con los hijos del nivel 4.2.1.1: elementos de las tres dates

day_received = ET.SubElement(date_received, "day")
month_received = ET.SubElement(date_received, "month")
year_received = ET.SubElement(date_received, "year")

day_rev_recd = ET.SubElement(date_rev_recd, "day")
month_rev_recd = ET.SubElement(date_rev_recd, "month")
year_rev_recd = ET.SubElement(date_rev_recd, "year")

day_accepted = ET.SubElement(date_accepted, "day")
month_accepted = ET.SubElement(date_accepted, "month")
year_accepted = ET.SubElement(date_accepted, "year")

day_received.text = articulo["day_received"]
month_received.text = articulo["month_received"]
year_received.text = articulo["year_received"]

day_rev_recd.text = articulo["day_rev_recd"]
month_rev_recd.text = articulo["month_rev_recd"]
year_rev_recd.text = articulo["year_rev_recd"]

day_accepted.text = articulo["day_accepted"]
month_accepted.text = articulo["month_accepted"]
year_accepted.text = articulo["year_accepted"]

#seguimos con los hijos del nivel 4.2: etiqueta permissions

permisos = articulo["permissions"]
permissions = ET.SubElement(article_meta, "permissions")

#seguimos con los hijos del nivel 4.2.1: hijos etiqueta permissions

copyright_statement = ET.SubElement(permissions, "copyright-statement")
copyright_year = ET.SubElement(permissions, "copyright-year")
copyright_holder = ET.SubElement(permissions, "copyright-holder")
copyright_statement.text = articulo["permissions"]["copyright_statement"]
copyright_year.text = articulo["permissions"]["copyright_year"]
copyright_holder.text = articulo["permissions"]["copyright_holder"]

#seguimos con los hijos del nivel 4.2.1.1: hijos etiqueta permissions

license = ET.SubElement(permissions, "license", attrib={
    "license-type": "open-access",
    "{http://www.w3.org/1999/xlink}href": "https://creativecommons.org/licenses/by/4.0/"
})

#seguimos con los hijos del nivel 4.2.1.1: hijos etiqueta license

license_ref = ET.SubElement(license, "{http://www.niso.org/schemas/ali/1.0/}license_ref")
license_p = ET.SubElement(license, "license-p")
license_ref.text = articulo["permissions"]["license_ref"]
license_p.text = articulo["permissions"]["license_p_text"]
ext_link = ET.SubElement(license_p, "ext-link", attrib={
    "ext-link-type": "uri",
    "{http://www.w3.org/1999/xlink}href": permisos["license_href"]
})
ext_link.text = permisos["license_p_link_text"]


#seguimos con los hijos del nivel 4.2: etiqueta abstract

abstract = ET.SubElement(article_meta, "abstract")
parrafo_abstract = ET.SubElement(abstract, "p")
parrafo_abstract.text = articulo["resumen_español"]
trans_abstract = ET.SubElement(article_meta, "trans-abstract", attrib={"xml:lang":"en"})
parrafo_trans_abstract = ET.SubElement(trans_abstract, "p")
parrafo_trans_abstract.text = articulo["abstract_ingles"]

kwd_group_español = ET.SubElement(article_meta, "kwd-group", attrib={"xml:lang":"es", "kwd-group-type":"author-keywords"})
kwd_group_ingles = ET.SubElement(article_meta, "kwd-group", attrib={"xml:lang":"en", "kwd-group-type":"author-keywords"})

#seguimos con los hijos del nivel 4.2.1: hijos etiqueta abstract

for palabras_clave in articulo["palabras_clave"]:
    kwd_español = ET.SubElement(kwd_group_español, "kwd")
    kwd_español.text = palabras_clave

for keywords in articulo["keywords"]:
    kwd_ingles = ET.SubElement(kwd_group_ingles, "kwd")
    kwd_ingles.text = keywords

#seguimos con los hijos del nivel 4.2: custom meta

custom_meta_group = ET.SubElement(article_meta, "custom-meta-group")
custom_meta_sumario = ET.SubElement(custom_meta_group, "custom-meta")
custom_meta_citar = ET.SubElement(custom_meta_group, "custom-meta")

#seguimos con los hijos del nivel 4.2.1: hijos etiqueta custom meta

meta_name_sumario = ET.SubElement(custom_meta_sumario, "meta-name")
meta_value_sumario = ET.SubElement(custom_meta_sumario, "meta-value")
meta_name_sumario.text = "sumario"
meta_value_sumario.text = articulo["sumario"]
meta_name_citar = ET.SubElement(custom_meta_citar, "meta-name")
meta_value_citar = ET.SubElement(custom_meta_citar, "meta-value")
meta_name_citar.text = "citar"
meta_value_citar.text = articulo["citar"]




#Para sacar el XML

import xml.dom.minidom

# Convertimos a formato pretty-print para poder visualizarlo
xml_str = xml.dom.minidom.parseString(ET.tostring(raiz, 'utf-8')).toprettyxml(indent="  ")

print(xml_str)


```
