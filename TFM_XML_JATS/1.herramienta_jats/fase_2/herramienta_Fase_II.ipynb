```python
# Biblioteca lxml

from lxml import etree as ET

# Namespaces con los que vamos a trabajar

NSMAP = {
    'xlink': "http://www.w3.org/1999/xlink",
    'mml': "http://www.w3.org/1998/Math/MathML",
    'xsi': "http://www.w3.org/2001/XMLSchema-instance",
    'ali': "http://www.niso.org/schemas/ali/1.0/",
    'oasis': "http://www.niso.org/standards/z39-96/ns/oasis-exchange/table"
}

# -- Diccionario --

# 1. Datos de la revista (información que no cambia entre artículos)
revista = {
    "journal_id": "ARIS",
    "journal_title": "Arte, individuo y sociedad",
    "abbrev_journal_title": "Art., indv. y soc.",
    "issn": "1131-5598",
    "issn_format": "electronic", # podría ser print si la revista se imprime, pero está en desuso
    "issn_l": "1131-5598",
    "publisher_name": "Universidad Complutense de Madrid",
    "publisher_country": "España",
    "website": "https://revistas.ucm.es/aris"
}

# 2. Datos del artículo (información específica de cada artículo)

articulo = {
    # Metadatos básicos
    "tipo_articulo": "research-article",  # opciones: research-article, review, editorial, letter, commentary, erratum, other
    "article_id": "10.5209/aris.78945",  # DOI completo
    "subject_section": "Artículo de investigación",
    "lang": "es",  # Idioma principal del artículo (define el xml:lang raíz y de ello depende el atributo de las palabras clave). Por defecto en español, pero podría ser "en", "pt", etc.

    # Títulos
    "article_title": "El impacto del arte urbano en la cohesión social",
    "trans_title": "The Impact of Urban Art on Social Cohesion",

    # Fechas importantes (formato DD, MM, YYYY)
    "publication_date": {"day": "15", "month": "03", "year": "2024"},
    "received_date": {"day": "02", "month": "08", "year": "2023"},
    "revised_date": {"day": "06", "month": "03", "year": "2024"},
    "accepted_date": {"day": "06", "month": "03", "year": "2024"},

    # Resúmenes
    "resumen_espanol": "Este estudio analiza cómo el arte urbano...",
    "abstract_ingles": "This study examines how urban art...",

    # Palabras clave
    "palabras_clave": ["arte urbano", "espacio público", "cohesión social", "cultura visual"],
    "keywords": ["urban art", "public space", "social cohesion", "visual culture"],

    # Contribuidores (lista de autores/colaboradores)
    "contributors": [
        {
            "contrib_id": "0000-0001-2345-6789",  # ORCID
            "surname": "García",
            "given_names": "María",
            "institution": "Universidad Complutense de Madrid",
            "country": "España",
            "country_code": "ES",  # Código ISO (atributo)
            "aff_id": "aff-1",
            "corresp_id": "cor-1",
            "email": "mgarcia@ucm.es",
            "roles": ["autoría", "conceptualización"]  # Opcional: CRediT taxonomy
        },
        {
            "contrib_id": "0000-0002-3456-7890",
            "surname": "López",
            "given_names": "Carlos",
            "institution": "Universidad Autónoma de Madrid",
            "country": "España",
            "country_code": "ES",  # Código ISO (atributo)
            "aff_id": "aff-2",
            "corresp_id": "cor-2",
            "email": "clopez@uam.es",
            "roles": ["metodología", "análisis formal"]
        },
    ]
}

# 3. PERMISOS Y LICENCIA (normalmente igual para todos los artículos, solo cambia el año)
permissions = {
    "copyright_statement": "Copyright © 2024, Universidad Complutense de Madrid",
    "copyright_year": "2024",
    "copyright_holder": "Universidad Complutense de Madrid",
    "license_href": "https://creativecommons.org/licenses/by/4.0/",
    "license_ref": "https://creativecommons.org/licenses/by/4.0/",
    "license_p_text": "Esta obra está bajo una licencia ",
    "license_p_link_text": "Creative Commons Attribution 4.0 International"
}

# -- Código --

# 1. Elemento raíz
article = ET.Element(
    "article",
    **{
        "article-type": articulo["tipo_articulo"],
        "dtd-version": "1.3"
    },
    nsmap=NSMAP
)
article.set("{http://www.w3.org/XML/1998/namespace}lang", articulo["lang"])

# 2. Front
front = ET.SubElement(article, "front")

# 3. Journal Meta
journal_meta = ET.SubElement(front, "journal-meta")
ET.SubElement(journal_meta, "journal-id", **{"journal-id-type": "publisher-id"}).text = revista["journal_id"]

title_group = ET.SubElement(journal_meta, "journal-title-group")
ET.SubElement(title_group, "journal-title", **{"specific-use": "original"}).text = revista["journal_title"]
ET.SubElement(title_group, "abbrev-journal-title", **{"abbrev-type": "publisher"}).text = revista["abbrev_journal_title"]

ET.SubElement(journal_meta, "issn", **{"publication-format": revista["issn_format"]}).text = revista["issn"]
ET.SubElement(journal_meta, "issn-l").text = revista["issn_l"]

publisher = ET.SubElement(journal_meta, "publisher")
ET.SubElement(publisher, "publisher-name").text = revista["publisher_name"]
ET.SubElement(ET.SubElement(publisher, "publisher-loc"), "country").text = revista["publisher_country"]

# 4. Article Meta
article_meta = ET.SubElement(front, "article-meta")

# 5 Elementos en orden:
ET.SubElement(article_meta, "article-id", **{"pub-id-type": "doi"}).text = articulo["article_id"]

# 4.2. Article Categories
subj_group = ET.SubElement(ET.SubElement(article_meta, "article-categories"), "subj-group", **{"subj-group-type": "section"})
ET.SubElement(subj_group, "subject").text = articulo["subject_section"]

# 6 Title Group
title_group = ET.SubElement(article_meta, "title-group")
ET.SubElement(title_group, "article-title", **{"{http://www.w3.org/XML/1998/namespace}lang": "es"}).text = articulo["article_title"]
ET.SubElement(ET.SubElement(title_group, "trans-title-group", **{"{http://www.w3.org/XML/1998/namespace}lang": "en"}), "trans-title").text = articulo["trans_title"]

# 7 Contrib Group
contrib_group = ET.SubElement(article_meta, "contrib-group")
for c in articulo["contributors"]:
    contrib = ET.SubElement(contrib_group, "contrib", **{"contrib-type": "author", "corresp": "yes" if "corresp_id" in c else "no"})
    ET.SubElement(contrib, "contrib-id", **{"contrib-id-type": "orcid"}).text = c["contrib_id"]
    name = ET.SubElement(contrib, "name")
    ET.SubElement(name, "surname").text = c["surname"]
    ET.SubElement(name, "given-names").text = c["given_names"]
    ET.SubElement(contrib, "xref", **{"ref-type": "aff", "rid": c["aff_id"]})
    if "corresp_id" in c:
        ET.SubElement(contrib, "xref", **{"ref-type": "corresp", "rid": c["corresp_id"]})

# 7.1. Afiliaciones
for c in articulo["contributors"]:
    aff = ET.SubElement(contrib_group, "aff", **{"id": c["aff_id"]})
    ET.SubElement(aff, "institution", **{"content-type": "original"}).text = c["institution"]
    country_elem = ET.SubElement(aff, "country", country=c["country_code"])  # Atributo ISO
    country_elem.text = c["country"]

# 7.2. Author Notes
author_notes = ET.SubElement(article_meta, "author-notes")
for c in [x for x in articulo["contributors"] if "corresp_id" in x]:
    corresp = ET.SubElement(author_notes, "corresp", **{"id": c["corresp_id"]})
    corresp.text = f"{c['given_names']} {c['surname']}"
    ET.SubElement(corresp, "email").text = c["email"]

# 7.3. Fecha de publicación (simplificado como pediste)
pub_date = ET.SubElement(article_meta, "pub-date",
                       **{"date-type": "pub",
                          "publication-format": "electronic",
                          "iso-8601-date": f"{articulo['publication_date']['year']}-{articulo['publication_date']['month']}-{articulo['publication_date']['day']}"})
ET.SubElement(pub_date, "day").text = articulo["publication_date"]["day"]
ET.SubElement(pub_date, "month").text = articulo["publication_date"]["month"]
ET.SubElement(pub_date, "year").text = articulo["publication_date"]["year"]

# 7.4. History
history = ET.SubElement(article_meta, "history")

# 7.5. Fecha recibido
date_data = articulo["received_date"]
date_elem = ET.SubElement(history, "date",
                        **{"date-type": "received",
                           "iso-8601-date": f"{date_data['year']}-{date_data['month']}-{date_data['day']}"})
ET.SubElement(date_elem, "day").text = date_data["day"]
ET.SubElement(date_elem, "month").text = date_data["month"]
ET.SubElement(date_elem, "year").text = date_data["year"]

# 7.6. Fecha revisión (usando revised_date como está en tu diccionario)
date_data = articulo["revised_date"]
date_elem = ET.SubElement(history, "date",
                        **{"date-type": "rev-recd",  # Pero mantenemos el tipo rev-recd que requiere JATS
                           "iso-8601-date": f"{date_data['year']}-{date_data['month']}-{date_data['day']}"})
ET.SubElement(date_elem, "day").text = date_data["day"]
ET.SubElement(date_elem, "month").text = date_data["month"]
ET.SubElement(date_elem, "year").text = date_data["year"]

# 7.7. Fecha aceptación
date_data = articulo["accepted_date"]
date_elem = ET.SubElement(history, "date",
                        **{"date-type": "accepted",
                           "iso-8601-date": f"{date_data['year']}-{date_data['month']}-{date_data['day']}"})
ET.SubElement(date_elem, "day").text = date_data["day"]
ET.SubElement(date_elem, "month").text = date_data["month"]
ET.SubElement(date_elem, "year").text = date_data["year"]

# 8. permissions (licencia y copyright)
perms_elem = ET.SubElement(article_meta, "permissions")
ET.SubElement(perms_elem, "copyright-statement").text = permissions["copyright_statement"]
ET.SubElement(perms_elem, "copyright-year").text = permissions["copyright_year"]
ET.SubElement(perms_elem, "copyright-holder").text = permissions["copyright_holder"]

license_elem = ET.SubElement(perms_elem, "license",
                           **{"license-type": "open-access",
                              "{http://www.w3.org/1999/xlink}href": permissions["license_href"]})
ET.SubElement(license_elem, "{http://www.niso.org/schemas/ali/1.0/}license_ref").text = permissions["license_ref"]

license_p = ET.SubElement(license_elem, "license-p")
license_p.text = permissions["license_p_text"]
ET.SubElement(license_p, "ext-link",
             **{"ext-link-type": "uri",
                "{http://www.w3.org/1999/xlink}href": permissions["license_href"]}).text = permissions["license_p_link_text"]

# 9. abstract y trans-abstract
abstract = ET.SubElement(article_meta, "abstract",
                       **{"{http://www.w3.org/XML/1998/namespace}lang": "es"})
ET.SubElement(abstract, "p").text = articulo["resumen_espanol"]

trans_abstract = ET.SubElement(article_meta, "trans-abstract",
                             **{"{http://www.w3.org/XML/1998/namespace}lang": "en"})
ET.SubElement(trans_abstract, "p").text = articulo["abstract_ingles"]

# 10. kwd-group (palabras clave)
for lang, kwds in [("es", "palabras_clave"), ("en", "keywords")]:
    # Atributos base (siempre incluye kwd-group-type)
    attrib = {"kwd-group-type": "author-keywords"}

    # Solo añade xml:lang si NO es el idioma principal del artículo
    if lang != articulo["lang"]:
        attrib["{http://www.w3.org/XML/1998/namespace}lang"] = lang

    kwd_group = ET.SubElement(article_meta, "kwd-group", **attrib)
    for kwd in articulo[kwds]:
        ET.SubElement(kwd_group, "kwd").text = kwd

# -- EXTRA: esto sirve para poder pasar la validación, al necesitar una estructura completa del XML JATS --

# 11 Cuerpo (body) de prueba para pasar la validación
body = ET.SubElement(article, "body")
sec = ET.SubElement(body, "sec")
title = ET.SubElement(sec, "title")
title.text = "Introducción"  # ¡Typo intencional para que notes que es de prueba!
p = ET.SubElement(sec, "p")
p.text = "Este es un párrafo de ejemplo para el cuerpo del artículo."

# 12 Sección final (back) de prueba para pasar la validación
back = ET.SubElement(article, "back")
ref_list = ET.SubElement(back, "ref-list")
ref = ET.SubElement(ref_list, "ref", id="B1")
citation = ET.SubElement(ref, "element-citation", **{"publication-type": "journal"})

# 12.1. Estructura semántica requerida con element-citation
person_group = ET.SubElement(citation, "person-group", **{"person-group-type": "author"})
name = ET.SubElement(person_group, "name")
ET.SubElement(name, "surname").text = "Apellido"
ET.SubElement(name, "given-names").text = "A"

ET.SubElement(citation, "year").text = "2023"
ET.SubElement(citation, "article-title").text = "Título del artículo"
ET.SubElement(citation, "source").text = "Revista"
ET.SubElement(citation, "volume").text = "1"
ET.SubElement(citation, "issue").text = "1"
ET.SubElement(citation, "fpage").text = "1"
ET.SubElement(citation, "lpage").text = "10"

# 13. Generar XML

DOCTYPE = '''<!DOCTYPE article PUBLIC "-//NLM//DTD JATS (Z39.96) Journal Publishing DTD with OASIS Tables with MathML3 v1.3 20210610//EN" "https://jats.nlm.nih.gov/publishing/1.3/JATS-journalpublishing-oasis-article1-3-mathml3.dtd">'''
XML_MODEL = '<?xml-model type="application/xml-dtd" href="http://jats.nlm.nih.gov/publishing/1.3/JATS-journalpublishing-oasis-article1-3-mathml3.dtd"?>'

xml_str = ET.tostring(
    article,
    encoding='UTF-8',  # Asegurar encoding en mayúsculas
    pretty_print=True,
    xml_declaration=False  # La añadiremos manualmente
)

with open('articulo_jats.xml', 'wb') as f:
    f.write(b'<?xml version="1.0" encoding="UTF-8"?>\n')
    f.write(XML_MODEL.encode('UTF-8') + b'\n')
    f.write(DOCTYPE.encode('UTF-8') + b'\n')
    f.write(xml_str)

# 14. Print para comprobar la descarga y enlace al validador

from IPython.display import display, HTML

display(HTML("""
<p>✅ Archivo <strong>articulo_jats.xml</strong> generado correctamente.</p>
<p>Valida tu XML JATS en: <a href="https://jats4r-validator.niso.org/" target="_blank">https://jats4r.org/validator</a></p>
"""))
```
