```python
!pip install weasyprint
from weasyprint import HTML, CSS

# Ruta al archivo HTML
html_path = '/content/181-192.html'

# Lee contenido del HTML
with open(html_path, 'r', encoding='utf-8') as f:
    html_content = f.read()

# CSS para forzar letra más pequeña
css = CSS(string="""
    body {
        font-size: 10pt !important;
    }
        h1, h2, h3, h4, h5, h6 {
        font-family: sans-serif !important;
    }
    figcaption, figcaption p, figcaption small,
figure p, figure p small {
    font-size: 8pt !important;
}
    sup {
        font-size: 70%;
        vertical-align: super;
        line-height: 0;
    }
""")

# Exportar como PDF e indicando la base_url para que se carguen bien las imágenes

output_pdf_path = '/content/181-192.pdf'
HTML(string=html_content, base_url='/content/').write_pdf(output_pdf_path, stylesheets=[css])

print("PDF generado en:", output_pdf_path)
```
